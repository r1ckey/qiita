```python
# カイ二乗検定のCSVを作る。CSV→DBeaver新規テーブル追加→Tableauに読み込み

# phase1 ライブラリのインポート
import pandas as pd
from scipy.stats import chi2_contingency
import os # 書き出しの指定


# phase2
# csv読み込み
df = pd.read_csv('C:/Users/barre/OneDrive/Documents/kaggle_01_AB/converted_csv/ab_test_preprocessed_202508191125.csv')

# 'group'と'converted'を軸に分割表（contingency table）を作成
# 行にグループ、列に結果（コンバージョン/非コンバージョン）
contingency_table = pd.crosstab(df['group'], df['converted'])

print("作成された分割表:")
print(contingency_table)

# 出力例:
# converted    0      1
# group
# control   97940  2500
# treatment 98060  2450

#Phase3 
# 分割表を使ってカイ二乗検定を実施
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nカイ二乗検定の結果:")
print(f"カイ二乗統計量: {chi2:.4f}")
print(f"P値: {p_value:.4f}")
print(f"自由度: {dof}")

#phase 4
# 結果を新しいデータフレームにまとめる
results_df = pd.DataFrame({
    'test_name': ['A/B_Test_Conversion'],
    'chi2_stat': [chi2],
    'p_value': [p_value],
    'degrees_of_freedom': [dof]
})

# CSVファイルとしてエクスポート
# index=Falseは、行番号をCSVに含めないための設定

# 読み込みファイルのパスを取得
input_file_path = 'C:/Users/barre/OneDrive/Documents/kaggle_01_AB/converted_csv/ab_test_preprocessed_202508191125.csv'
# 読み込みファイルのあるディレクトリを取得
output_dir = os.path.dirname(input_file_path)

# 出力パスを作成
output_file_path = os.path.join(output_dir, 'chi2_results.csv')

# 結果をCSVファイルとしてエクスポート
results_df.to_csv(output_file_path, index=False)
print("\n結果が 'chi2_results.csv' としてエクスポートされました。")

# jupyternotebook保存先の確認
import os
os.getcwd()
```
